```python
import pandas as pd
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
data = pd.read_csv("AMZN_DATA_WITH_NEWS_NO_SENTIMENT.csv")
data.head()
```

```python
nltk.download('vader_lexicon')

stock_sentiment = {
    'surges': 15,
    'rallies': 12,
    'beats': 10,
    'climbs': 8,
    'tops': 6,
    'misses': -8,
    'drops': -10,
    'plunges': -20,
    'warns': -15,
    'crashes': -25,
}

vader = SentimentIntensityAnalyzer()
vader.lexicon.update(stock_sentiment)
```

```python
scores = [vader.polarity_scores (titles) for titles in data['article-title'].values]
```

```python
scores_df = pd.DataFrame(scores)
data = pd.concat([data, scores_df], axis = 1)
```
