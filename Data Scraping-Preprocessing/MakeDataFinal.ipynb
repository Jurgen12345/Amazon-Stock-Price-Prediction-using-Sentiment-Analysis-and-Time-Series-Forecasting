```python
import pandas as pd

historical_data = pd.read_csv("AMZN_HISTORICAL_DATA.csv")
historical_data.head()
```

```python
news_articles = pd.read_csv("AMAZON_NEWS_HEADLINES_FINAL.csv")
news_articles.head()
print(len(news_articles.index))
```

```python
news_articles = news_articles.rename(columns={"0":"article-title"})
news_articles.head()
```

```python
news_articles["article-title"] = news_articles["article-title"].str.split('•')
```

```python
news_articles.head()
```

```python
news_articles['article-title'] = news_articles['article-title'].astype(str)
news_articles['article-title'] = news_articles['article-title'].str.replace('[', '', regex = False).str.replace(']', '', regex = False)
```

```python
news_articles["article-title"] = news_articles['article-title'].str.replace(r"\n", "")
news_articles.head(10)
```

```python
def clean_title(title):
    title = str(title)
    if len(title) >= 5:
      if title[-5] == ',':
        return title[:-5]
    return title
```

```python
news_articles['article-title'] = news_articles['article-title'].apply(clean_title)

news_articles.head(20)
```

```python
news_articles['article-date'] = news_articles['article-title'].str[-13:-1]
news_articles.head(50)
```

```python
historical_data['Date'] = pd.to_datetime(historical_data['Date'].str[:10]).dt.strftime('%Y-%m-%d')
historical_data.head()
```

```python
pd.set_option('display.max_rows', None)
```

```python
mask = ~news_articles['article-date'].str.contains(r'\n', regex = True, na = False)
news_articles.loc[mask, 'article-date'] = pd.to_datetime(news_articles.loc[mask, 'article-date'], errors ='coerce').dt.strftime("%Y-%m-%d")
```

```python
news_articles['article-title'] = news_articles['article-title'].astype(str)
news_articles['article-title'] = news_articles['article-title'].str.replace('[', '', regex = False).str.replace(']', '', regex = False)
news_articles.head()
```

```python
news_articles['article-date'] = pd.to_datetime(news_articles['article-date'], errors ='coerce').dt.strftime('%Y-%m-%d')
news_articles.head(50)
```

```python
news_articles = news_articles.dropna(subset=['article-date'])
news_articles
```

```python
len(news_articles.index)
```

```python
news_articles_grouped = news_articles.groupby('article-date')['article-title'].apply(list).reset_index()
news_articles_grouped = news_articles_grouped.rename(columns={"article-date":'Date'})
news_articles_grouped.head(10)
```

```python
final_data = pd.merge(historical_data, news_articles_grouped,on="Date", how='inner')
final_data.head()
```

```python
print(len(final_data))
```

```python
final_data['article-title'] = final_data['article-title'].astype(str)
final_data['article-title'] = final_data['article-title'].str.replace('[','', regex = False).str.replace(']','' ,regex = False)
final_data.head()
```

```python
final_data.to_csv('AMZN_DATA_WITH_NEWS_NO_SENTIMENT.csv',index = False)
```
